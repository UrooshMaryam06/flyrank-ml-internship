## 5. Why ML beats a fixed rule here

A fixed rule like "flag pages with days_since_last_update >= 180" only
looks at one signal at a time. But whether a page is actually worth
reviewing depends on several signals moving together and interacting
differently across content types: staleness, demand level, position,
CTR gap, and engagement. Hand-writing every useful combination of
thresholds across those signals isn't realistic. The starter pipeline
already shows the gap concretely: the hand rule hits Precision@50 of
0.24, the random forest hits 0.74 on the identical task — the model
finds combinations a human-written rule doesn't catch.

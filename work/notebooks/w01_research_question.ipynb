"which pages should we look at first?" into a ranked, evidence-backed queue
instead of a gut call, using observable signals (impressions, position,
staleness, trend) rather than the product's own decision flags. I picked it

Decision this improves: which declining/stale pages a content reviewer looks
at first, out of thousands, with limited weekly review capacity.

Who acts on it, and what they do: a content strategist or SEO lead, who
decides whether to refresh, expand, protect, or leave a page alone.

Cost of a wrong call: two ways to be wrong, and they cost differently.
Flagging a healthy page wastes review time. Missing a genuinely declining,
high-demand page means it keeps losing visibility unnoticed. A ranked queue
with reason codes lets a reviewer sanity-check before acting, so mistakes
stay cheap.

Why this isn't just "train a model": a plain rule (e.g. "flag anything with
days_since_last_update >= 180") already exists as the starter baseline, and
it works passably. But the real pattern is several signals tangled together
- staleness, demand, position, CTR, engagement - shifting differently for
different content types. That's too messy to hand-write reliably, which is
exactly where a learned ranking earns its place. The starter pipeline's own
numbers make this concrete: the hand rule hits Precision@50 of 0.24; the
random forest hits 0.74 on the same task.

## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
import pandas as pd

df = pd.read_csv("data/raw/content_refresh_anonymized.csv")
filt = df[(df["impressions_90d"] > 0) & (df["content_age_days"] >= 90)]

declining_pct = (filt["trend_direction"] == "down").mean()
declining_with_demand = ((filt["trend_direction"] == "down") & (filt["impressions_90d"] >= 100)).sum()

print(f"Rows after basic quality filter: {len(filt):,}")
print(f"Share currently trending down: {declining_pct:.1%}")
print(f"Declining pages with real demand (impressions_90d >= 100): {declining_with_demand:,}")
```

## 4. Careful words: what I can and can't claim

Can say: which pages show observed decline, staleness, or CTR/position
weakness right now, ranked so a reviewer spends limited time on the most
promising candidates first. This is decision-support, not proof.

Cannot say: that a refresh causes recovery (no experiment run), that any
result reveals a Google ranking factor, or that a model probability is a
guarantee. "Down" here is a current-window proxy label, not a validated
future outcome — a stronger capstone version would use a true future window
(prior 90 days → next 30 days decline).
